```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Un dataframe sencillo

```python
df = pd.DataFrame({
    "mes": ["Enero", "Febrero", "Marzo", "Abril"],
    "temperatura": [4.2, 5.8, 9.1, 12.4],
    "precipitacion": [45, 38, 52, 60],
})
df
```

## 2. Verlo en la consola / como salida de celda
Al ser la última expresión de la celda, Jupyter lo muestra automáticamente como tabla (sin necesitar `print`).

## 3. Data Viewer
En VS Code: pestaña **Jupyter: Variables** (arriba del notebook), o clic derecho sobre `df` en esa lista → **View Value in Data Viewer**.

## 4. Alguna transformación básica

```python
df["temperatura_f"] = df["temperatura"] * 9 / 5 + 32
df
```

## 5. Un resumen rápido

```python
df.describe()
```

## 6. Gráfico para ver el panel de salida

```python
plt.bar(df["mes"], df["temperatura"])
plt.title("Temperatura por mes")
plt.ylabel("ºC")
plt.show()
```
